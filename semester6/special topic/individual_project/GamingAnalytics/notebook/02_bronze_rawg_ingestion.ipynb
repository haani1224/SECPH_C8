```python
games = []

for page in range(1, 20):

    params = {
        "key": API_KEY,
        "page": page,
        "page_size": 40
    }

    response = requests.get(
        url,
        params=params
    )

    result = response.json()

    games.extend(
        result["results"]
    )

print(
    "Total games collected:",
    len(games)
)
```

```python
import json

with open(
    "../data/bronze/rawg_games.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        games,
        file,
        indent=4
    )

print("RAWG bronze saved")
```
